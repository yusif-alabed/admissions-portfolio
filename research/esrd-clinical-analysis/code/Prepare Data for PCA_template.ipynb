# Isolate numerical columns for PCA

```python
# Step 4

# Get the names of only the numerical columns
numerical_cols = df.select_dtypes(include=['number']).columns
numerical_cols

# Create a subset of containing only numerical columns to remove any label columns
df = df[numerical_cols]
df
```

# Find outliers in ALL columns that are numerical (Using the IQR Rule of 1.5)

```python
# Step 5

# Define the function
def find_outliers(df):
    outliers = {} 
    all_outlier_indices = set()

    for col in df.select_dtypes(include='number').columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1 
        lower_bound = Q1 - 1.5 * IQR 
        upper_bound = Q3 + 1.5 * IQR 

        # Get indices of outliers for this column
        outlier_indices = df[(df[col] < lower_bound) | (df[col] > upper_bound)].index.tolist()
        outliers[col] = outlier_indices

        # Add to the full set of outlier indices
        all_outlier_indices.update(outlier_indices)

    # Create DataFrame of rows that contain any outlier
    outlier_rows = df.loc[sorted(all_outlier_indices)]

    return outliers, outlier_rows

# Call the function
location, actual_rows = find_outliers(df)
location
```

# Write/Paste in names of columns that HAVE OUTLIERS

```python
# Step 6

# Paste in column names that have outliers (these need RobustScaler)
hasOutliers = [ '3-Methylhistidine', 'Androsterone sulfate-2', 'Androsterone sulfate-3', 'Androsterone sulfate-4', 'chenodeoxycholic acid', 'Cholic acid', 'Creatinine', ]
hasOutliers
```

```python
# Extract which columns do not contain outliers (these need StandardScaler)
column_names = df.columns # Get all column names
noOutliers = [col for col in column_names if col not in hasOutliers]
noOutliers
```

# Standardize "noOutliers" via StandardScaler

```python
# Step 7

df = df.copy()  # Ensures df is a standalone DataFrame, not a slice

# Instantiate scaler
meanSD_scaler = StandardScaler()

# Convert numerical columns to float64 (in place)
df[noOutliers] = df[noOutliers].astype('float64')

# Apply scaling
df[noOutliers] = meanSD_scaler.fit_transform(df[noOutliers])

df
```

# Standardize "hasOutliers" via RobustScaler

```python
# Step 8

df = df.copy()  # Ensures df is a standalone DataFrame, not a slice

# Instantiate scaler
medIQR_scaler = RobustScaler()

# Convert numerical columns to float64 (in place)
df[hasOutliers] = df[hasOutliers].astype('float64')

# Apply scaling
df[hasOutliers] = medIQR_scaler.fit_transform(df[hasOutliers])

df
```

```python
# Step 9

# Combine the label columns with the standardized numerical columns
combined_df = pd.concat([labels_df, df], axis=1)
```

```python
# Step 10

# Create new output filename
fileName = base_name + "_standardized4PCA.csv"

# Save the DataFrame
combined_df.to_csv(fileName, index=False)
```
